```python
import ollama
```

```python
#generation
response = ollama.generate(model="llama3", prompt="What is the capital of France?")
```

```python
print(response)
```

```python
print(response.response)
```

```python
#stream response--used if we want to see how o/p is generated
for r in ollama.generate(model="llama3", prompt="What are the places to explore in hyderabad?",stream=True):
    print(r.response, end="")
```

```python
#generation-by passing system i/p
response = ollama.generate(model="llama3", prompt="What is the capital of France?",system="You are a funny assistant u explain things in funny way.")
```

```python
print(response)
```

```python
#to change parameterss of model
res=ollama.generate(model="llama3", prompt="Why is the ocean blue?",options={"temperature": 0.8,
                                                                             "top_p": 0.5,
                                                                             "top_k": 40} )
print(res.response)
```

```python
#integrating with langchain
#for chat conversation
from langchain_ollama import ChatOllama

#initialise the model
llm = ChatOllama(model="llama3",temperature=0,top_k=30)

response=llm.invoke("Explain the concept of dejavu in 4 sentences.")
print(response.content)
```

```python
#for non chat execution
from langchain_ollama import OllamaLLM

llm = OllamaLLM(model="llama3")

response = llm.invoke("what is the capital of France?")
print(response)
```

```python
from langchain_ollama import OllamaEmbeddings

embeddings = OllamaEmbeddings(model="llama3")
# ResponseError: This model does not support embeddings
#to get embedding of a text
text = "This is a sample text for generating embeddings."
print(embeddings.embed_query(text))
```
