```python
from langgraph.graph import StateGraph,START,END
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI
from typing import TypedDict
```

```python
load_dotenv()
model = ChatGoogleGenerativeAI(model = 'gemini-2.5-flash')
```

```python
#define state and functions

class State(TypedDict):
    topic : str
    outline : str
    blog : str
    evaluated_score : str

def gen_outline(state : State)->State:
    topic = state['topic']
    prompt = f"generate detailed outline for the topic{topic}"
    state['outline'] = model.invoke(prompt).content
    return state

def gen_blog(state : State)->State:
    outline = state['outline']
    prompt = f"Generate blog from the provided outline{outline}"
    state['blog'] = model.invoke(prompt).content
    return state

def gen_eval_score(state:State)->State :
    blog = state['blog']
    outline = state['outline']
    prompt = f'By refering this this outline\n{outline} evaluate the score for the generated blog{blog} out of 10'
    state['evaluated_score'] = model.invoke(prompt).content
    return state
```

```python
#define graph
graph = StateGraph(State)

# nodes
graph.add_node('create_outline', gen_outline)
graph.add_node('create_blog', gen_blog)
graph.add_node('evaluate_score', gen_eval_score)


# edges
graph.add_edge(START, 'create_outline')
graph.add_edge('create_outline', 'create_blog')
graph.add_edge( 'create_blog','evaluate_score')
graph.add_edge('evaluate_score', END)

workflow = graph.compile()
```

```python
intial_state = {'topic': 'Rise of AI in India'}
final_state = workflow.invoke(intial_state)
print(final_state)
```
